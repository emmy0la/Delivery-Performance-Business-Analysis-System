#### 1. Problem Understanding

##### Business Problem

SmartLogistics needs a simple system for tracking customer deliveries.

The system should:

- Store delivery information.
- Validate delivery records.
- Determine delivery performance.
- Determine delivery priority.
- Recommend an action.
- Process multiple deliveries.
- Generate a delivery report.
- Produce summary statistics.

##### Delivery Performance

- Cancelled → Cancelled
- Delivered → Completed
- Days Since Order <= Expected Delivery Days → On Time
- Days Since Order > Expected Delivery Days → Delayed

##### Delivery Priority

- Delayed + 2 or more attempts → Urgent
- Delayed + at least 1 attempt → High
- On Time + at least 1 attempt → Normal
- On Time + 0 attempts → Low

The goal is to break the problem into smaller reusable functions.

```python
# ============================================
# VALIDATION
# ============================================
# Validates each piece of delivery information
# Stores results as Boolean values (True/False)
# ============================================

# Accepted delivery statuses
valid_statuses = [
    "Order Received",
    "Processing",
    "Dispatched",
    "In Transit",
    "Delivered",
    "Delayed",
    "Cancelled"
]


def validate_delivery_record(delivery):
    """
    Validate one delivery record.
    Returns 8 Boolean values.
    """
    
    # Check text fields are not empty
    valid_id = delivery["delivery_id"] != ""
    valid_name = delivery["customer_name"] != ""
    valid_location = delivery["delivery_location"] != ""
    
    # Check status is accepted
    valid_status = delivery["delivery_status"] in valid_statuses
    
    # Check numeric values are valid
    valid_days = delivery["days_since_order"] >= 0
    valid_expected_days = delivery["expected_delivery_days"] > 0
    valid_weight = delivery["package_weight"] > 0
    valid_attempts = delivery["delivery_attempts"] >= 0
    
    return (
        valid_id,
        valid_name,
        valid_location,
        valid_status,
        valid_days,
        valid_expected_days,
        valid_weight,
        valid_attempts
    )


def validate_delivery(
    valid_id,
    valid_name,
    valid_location,
    valid_status,
    valid_days,
    valid_expected_days,
    valid_weight,
    valid_attempts
):
    """
    Determine whether the complete record is valid.
    Uses logical operator 'and' to combine all checks.
    """
    
    if (
        valid_id
        and valid_name
        and valid_location
        and valid_status
        and valid_days
        and valid_expected_days
        and valid_weight
        and valid_attempts
    ):
        return "VALID"
    else:
        return "INVALID"


# ============================================
# TEST VALIDATION ON DELIVERY 1
# ============================================

# Get validation results
validation_results = validate_delivery_record(delivery_1)

# Unpack the results
(
    valid_id,
    valid_name,
    valid_location,
    valid_status,
    valid_days,
    valid_expected_days,
    valid_weight,
    valid_attempts
) = validation_results

# Determine final validation status
validation_status = validate_delivery(
    valid_id,
    valid_name,
    valid_location,
    valid_status,
    valid_days,
    valid_expected_days,
    valid_weight,
    valid_attempts
)

print("=" * 50)
print("   VALIDATION TEST - DELIVERY 1")
print("=" * 50)
print("Delivery ID:", valid_id)
print("Customer Name:", valid_name)
print("Delivery Location:", valid_location)
print("Delivery Status:", valid_status)
print("Days Since Order:", valid_days)
print("Expected Delivery Days:", valid_expected_days)
print("Package Weight:", valid_weight)
print("Delivery Attempts:", valid_attempts)
print("-" * 50)
print("Delivery Record:", validation_status)
print("=" * 50)
print()
```

```python
# ============================================
# DELIVERY PERFORMANCE
# ============================================
# Determines if delivery is:
# - Cancelled
# - Completed
# - On Time
# - Delayed
# ============================================

def check_delivery_performance(
    delivery_status,
    days_since_order,
    expected_delivery_days
):
    """
    Determine delivery performance.
    Uses if, elif, and else statements.
    """
    
    # Cancelled delivery
    if delivery_status == "Cancelled":
        return "Cancelled"
    
    # Delivered delivery
    elif delivery_status == "Delivered":
        return "Completed"
    
    # Active delivery within expected period
    elif days_since_order <= expected_delivery_days:
        return "On Time"
    
    # Active delivery past expected period
    else:
        return "Delayed"


# ============================================
# TEST PERFORMANCE ON DELIVERY 1
# ============================================

performance = check_delivery_performance(
    delivery_1["delivery_status"],
    delivery_1["days_since_order"],
    delivery_1["expected_delivery_days"]
)

print("=" * 50)
print("   PERFORMANCE TEST - DELIVERY 1")
print("=" * 50)
print("Status:", delivery_1["delivery_status"])
print("Days Since Order:", delivery_1["days_since_order"])
print("Expected Delivery Days:", delivery_1["expected_delivery_days"])
print("-" * 50)
print("Performance:", performance)
print("=" * 50)
print()
```

```python
# ============================================
# DELIVERY PRIORITY
# ============================================
# Determines if delivery is:
# - Urgent
# - High
# - Normal
# - Low
# Uses logical operator 'and' to combine conditions
# ============================================

def determine_delivery_priority(
    delivery_performance,
    days_since_order,
    expected_delivery_days,
    delivery_attempts
):
    """
    Determine delivery priority.
    Uses logical operators and conditional statements.
    """
    
    # Delayed + 2 or more attempts = Urgent
    if (
        delivery_performance == "Delayed"
        and delivery_attempts >= 2
    ):
        return "Urgent"
    
    # Delayed + at least 1 attempt = High
    elif (
        delivery_performance == "Delayed"
        and delivery_attempts >= 1
    ):
        return "High"
    
    # On Time + at least 1 attempt = Normal
    elif (
        delivery_performance == "On Time"
        and delivery_attempts >= 1
    ):
        return "Normal"
    
    # On Time + 0 attempts = Low
    else:
        return "Low"


# ============================================
# TEST PRIORITY ON DELIVERY 2
# ============================================

performance_2 = check_delivery_performance(
    delivery_2["delivery_status"],
    delivery_2["days_since_order"],
    delivery_2["expected_delivery_days"]
)

priority_2 = determine_delivery_priority(
    performance_2,
    delivery_2["days_since_order"],
    delivery_2["expected_delivery_days"],
    delivery_2["delivery_attempts"]
)

print("=" * 50)
print("   PRIORITY TEST - DELIVERY 2")
print("=" * 50)
print("Status:", delivery_2["delivery_status"])
print("Days Since Order:", delivery_2["days_since_order"])
print("Expected Delivery Days:", delivery_2["expected_delivery_days"])
print("Delivery Attempts:", delivery_2["delivery_attempts"])
print("-" * 50)
print("Performance:", performance_2)
print("Priority:", priority_2)
print("=" * 50)
print()
```

```python
# ============================================
# RECOMMENDED ACTION
# ============================================
# Maps priority to recommended action
# ============================================

def recommend_delivery_action(delivery_priority):
    """
    Recommend an action based on delivery priority.
    Uses if, elif, and else.
    """
    
    if delivery_priority == "Urgent":
        return "Contact Customer and Escalate Delivery"
    
    elif delivery_priority == "High":
        return "Investigate Delivery Delay"
    
    elif delivery_priority == "Normal":
        return "Continue Delivery Process"
    
    else:
        return "Monitor Delivery"


# ============================================
# TEST ACTION ON DELIVERY 2
# ============================================

action_2 = recommend_delivery_action(priority_2)

print("=" * 50)
print("   ACTION TEST - DELIVERY 2")
print("=" * 50)
print("Priority:", priority_2)
print("-" * 50)
print("Recommended Action:", action_2)
print("=" * 50)
print()
```

```python
# ============================================
# GENERATE DELIVERY REPORT
# ============================================
# Combines all functions to produce structured report
# ============================================

def generate_delivery_report(delivery):
    """
    Generate a structured delivery report.
    Uses global variable 'company_name'.
    Creates local variable 'report_title'.
    """
    
    # Local variable (exists only inside this function)
    report_title = "DELIVERY STATUS REPORT"
    
    # Calculate performance
    performance = check_delivery_performance(
        delivery["delivery_status"],
        delivery["days_since_order"],
        delivery["expected_delivery_days"]
    )
    
    # Calculate priority
    priority = determine_delivery_priority(
        performance,
        delivery["days_since_order"],
        delivery["expected_delivery_days"],
        delivery["delivery_attempts"]
    )
    
    # Calculate action
    action = recommend_delivery_action(priority)
    
    # Validate delivery
    (
        valid_id,
        valid_name,
        valid_location,
        valid_status,
        valid_days,
        valid_expected_days,
        valid_weight,
        valid_attempts
    ) = validate_delivery_record(delivery)
    
    validation_status = validate_delivery(
        valid_id,
        valid_name,
        valid_location,
        valid_status,
        valid_days,
        valid_expected_days,
        valid_weight,
        valid_attempts
    )
    
    # Display report
    print("=" * 50)
    print(f"       {company_name.upper()}")
    print(f"       {report_title}")
    print("=" * 50)
    
    print()
    print("Delivery ID:", delivery["delivery_id"])
    print("Customer Name:", delivery["customer_name"])
    print("Delivery Location:", delivery["delivery_location"])
    
    print()
    print("-" * 40)
    print("DELIVERY INFORMATION")
    print("-" * 40)
    
    print("Status:", delivery["delivery_status"])
    print("Days Since Order:", delivery["days_since_order"], "days")
    print("Expected Delivery:", delivery["expected_delivery_days"], "days")
    print("Package Weight:", delivery["package_weight"], "kg")
    print("Delivery Attempts:", delivery["delivery_attempts"])
    
    print()
    print("-" * 40)
    print("DELIVERY ANALYSIS")
    print("-" * 40)
    
    print("Performance:", performance)
    print("Priority:", priority)
    print("Recommended Action:", action)
    
    print()
    print("-" * 40)
    print("VALIDATION")
    print("-" * 40)
    
    print("Delivery Record:", validation_status)
    
    print("=" * 50)
    print()


# ============================================
# TEST REPORT ON DELIVERY 1
# ============================================

generate_delivery_report(delivery_1)
```

```python
# ============================================
# PROCESS ALL DELIVERIES WITH FOR LOOP
# ============================================
# Uses for loop to process all six records
# Without repeating the same code
# ============================================

print("=" * 50)
print("   PROCESSING ALL DELIVERIES")
print("=" * 50)
print()

for delivery in deliveries:
    
    generate_delivery_report(delivery)
```

```python
# ============================================
# WHILE LOOP: Validate Delivery Attempts
# ============================================
# Continues asking until valid input is entered
# ============================================

print("=" * 50)
print("   WHILE LOOP: Validate Delivery Attempts")
print("=" * 50)
print()

# Start with invalid value
delivery_attempts = -1

# Loop continues while value is invalid
while delivery_attempts < 0:
    
    # Get user input
    user_input = input("Enter number of delivery attempts: ")
    delivery_attempts = int(user_input)
    
    # Check if value is valid
    if delivery_attempts < 0:
        print("Delivery attempts cannot be negative.")
        print("Please try again.")
        print()
    
    else:
        print("Delivery attempts accepted.")
        print()

print("=" * 50)
print()
```

```python
# ============================================
# WHILE LOOP: Validate Delivery Status
# ============================================
# Continues asking until valid status is entered
# ============================================

print("=" * 50)
print("   WHILE LOOP: Validate Delivery Status")
print("=" * 50)
print()

# Start with invalid value
delivery_status = ""

# Loop continues while value is invalid
while delivery_status not in valid_statuses:
    
    # Get user input
    delivery_status = input("Enter delivery status: ")
    
    # Check if value is valid
    if delivery_status not in valid_statuses:
        print("Invalid delivery status.")
        print("Please enter a valid delivery status.")
        print()
    
    else:
        print("Delivery status accepted.")
        print()

print("=" * 50)
print()
```

```python
# ============================================
# INFINITE WHILE LOOP
# ============================================
# Uses while True to create controlled infinite loop
# Uses break to exit when user says "no"
# ============================================

print("=" * 50)
print("   INFINITE WHILE LOOP")
print("=" * 50)
print()

# Controlled infinite loop
while True:
    
    # Ask user if they want to continue
    response = input("Do you want to record another delivery? (yes/no): ")
    response = response.lower()
    
    if response == "yes":
        print("Continue recording delivery.")
        print()
    
    elif response == "no":
        print("Stopping delivery entry.")
        print()
        break
    
    else:
        print("Please enter yes or no.")
        print()

print("=" * 50)
print()
```

```python
# ============================================
# BONUS ANALYSIS
# ============================================
# Processes all deliveries and calculates statistics
# ============================================

print("=" * 50)
print("   BONUS ANALYSIS")
print("=" * 50)
print()

# Initialize counters
total_deliveries = 0
on_time_count = 0
delayed_count = 0
completed_count = 0
cancelled_count = 0

urgent_count = 0
high_count = 0
normal_count = 0
low_count = 0

escalation_count = 0
no_attempt_count = 0

longest_delivery = deliveries[0]
highest_attempt_delivery = deliveries[0]
heaviest_delivery = deliveries[0]

# Process each delivery
for delivery in deliveries:
    
    # Get performance
    performance = check_delivery_performance(
        delivery["delivery_status"],
        delivery["days_since_order"],
        delivery["expected_delivery_days"]
    )
    
    # Get priority
    priority = determine_delivery_priority(
        performance,
        delivery["days_since_order"],
        delivery["expected_delivery_days"],
        delivery["delivery_attempts"]
    )
    
    # Count deliveries
    total_deliveries += 1
    
    # Count performance
    if performance == "On Time":
        on_time_count += 1
    
    elif performance == "Delayed":
        delayed_count += 1
    
    elif performance == "Completed":
        completed_count += 1
    
    elif performance == "Cancelled":
        cancelled_count += 1
    
    # Count priority
    if priority == "Urgent":
        urgent_count += 1
        escalation_count += 1
    
    elif priority == "High":
        high_count += 1
    
    elif priority == "Normal":
        normal_count += 1
    
    elif priority == "Low":
        low_count += 1
    
    # Count no attempts
    if delivery["delivery_attempts"] == 0:
        no_attempt_count += 1
    
    # Find longest open delivery
    if delivery["days_since_order"] > longest_delivery["days_since_order"]:
        longest_delivery = delivery
    
    # Find highest attempts
    if delivery["delivery_attempts"] > highest_attempt_delivery["delivery_attempts"]:
        highest_attempt_delivery = delivery
    
    # Find heaviest package
    if delivery["package_weight"] > heaviest_delivery["package_weight"]:
        heaviest_delivery = delivery

# Calculate percentage
on_time_percentage = (on_time_count / total_deliveries) * 100

# Display analysis
print("DELIVERY PERFORMANCE")
print("-" * 40)
print("Total Deliveries:", total_deliveries)
print("On Time Deliveries:", on_time_count)
print("Delayed Deliveries:", delayed_count)
print("Completed Deliveries:", completed_count)
print("Cancelled Deliveries:", cancelled_count)
print()

print("DELIVERY PRIORITY")
print("-" * 40)
print("Urgent Deliveries:", urgent_count)
print("High Priority Deliveries:", high_count)
print("Normal Priority Deliveries:", normal_count)
print("Low Priority Deliveries:", low_count)
print()

print("ADDITIONAL INSIGHTS")
print("-" * 40)
print("Deliveries Requiring Escalation:", escalation_count)
print("Deliveries With No Attempts:", no_attempt_count)
print()

print("LONGEST OPEN DELIVERY")
print("-" * 40)
print("Delivery ID:", longest_delivery["delivery_id"])
print("Customer:", longest_delivery["customer_name"])
print("Days Since Order:", longest_delivery["days_since_order"], "days")
print()

print("HIGHEST DELIVERY ATTEMPTS")
print("-" * 40)
print("Delivery ID:", highest_attempt_delivery["delivery_id"])
print("Customer:", highest_attempt_delivery["customer_name"])
print("Attempts:", highest_attempt_delivery["delivery_attempts"])
print()

print("HIGHEST PACKAGE WEIGHT")
print("-" * 40)
print("Delivery ID:", heaviest_delivery["delivery_id"])
print("Customer:", heaviest_delivery["customer_name"])
print("Weight:", heaviest_delivery["package_weight"], "kg")
print()

print("DELIVERY REQUIRING IMMEDIATE ATTENTION")
print("-" * 40)
# Find first urgent delivery
for delivery in deliveries:
    performance = check_delivery_performance(
        delivery["delivery_status"],
        delivery["days_since_order"],
        delivery["expected_delivery_days"]
    )
    priority = determine_delivery_priority(
        performance,
        delivery["days_since_order"],
        delivery["expected_delivery_days"],
        delivery["delivery_attempts"]
    )
    if priority == "Urgent":
        print("Delivery ID:", delivery["delivery_id"])
        print("Customer:", delivery["customer_name"])
        break

print()
print("ON TIME PERCENTAGE")
print("-" * 40)
print("On Time Percentage:", round(on_time_percentage, 2), "%")
print()
```
